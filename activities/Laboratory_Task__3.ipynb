Backward Propagation & Parameter UpdateOutput Gradient ($d_o$):$$d_o = (\hat{y} - y) \cdot f'(Z_o) = (0.08 - 1.0) \cdot 1 = -0.92$$Gradients for Output Weights ($dW_o$) & Biases ($d\theta_o$):$$dW_o = A_h^T \cdot d_o = [0.0, 0.1]^T \cdot (-0.92) = [0.0, -0.092]$$$$d\theta_o = d_o = -0.92$$Hidden Layer Gradient ($d_h$):$$d_h = (d_o \cdot W_o^T) \odot f'(Z_h)$$Since $Z_h = [-0.7, 0.1]$, the ReLU derivative $f'(Z_h) = [0, 1]$.$$d_o \cdot W_o^T = -0.92 \cdot [-0.3, -0.2] = [0.276, 0.184]$$$$d_h = [0.276, 0.184] \odot [0, 1] = [0.0, 0.184]$$Gradients for Hidden Weights ($dW_h$) & Biases ($d\theta_h$):$$dW_h = x^T \cdot d_h = \begin{bmatrix} 1 \\ 0 \\ 1 \end{bmatrix} \begin{bmatrix} 0.0 & 0.184 \end{bmatrix} = \begin{bmatrix} 0.0 & 0.184 \\ 0.0 & 0.0 \\ 0.0 & 0.184 \end{bmatrix}$$$$d\theta_h = d_h = [0.0, 0.184]$$Weight Updates ($\eta = 0.001$):$$W_o^{(new)} = W_o - \eta \cdot dW_o = [-0.3, -0.2] - 0.001 \cdot [0.0, -0.092] = [-0.3, -0.199908]$$$$W_h^{(new)} = W_h - \eta \cdot dW_h = \begin{bmatrix} 0.2 & -0.3 \\ 0.4 & 0.1 \\ -0.5 & 0.2 \end{bmatrix} - 0.001 \begin{bmatrix} 0.0 & 0.184 \\ 0.0 & 0.0 \\ 0.0 & 0.184 \end{bmatrix} = \begin{bmatrix} 0.2 & -0.300184 \\ 0.4 & 0.100000 \\ -0.5 & 0.199816 \end{bmatrix}$$

```python
import torch
import numpy as np

# Manual implementation matching the formulas
x = np.array([1.0, 0.0, 1.0])
y = 1.0
lr = 0.001

W_h = np.array([
    [0.2, -0.3],
    [0.4, 0.1],
    [-0.5, 0.2]
])
theta_h = np.array([-0.4, 0.2])

W_o = np.array([-0.3, -0.2])
theta_o = 0.07

# Forward pass
Z_h = np.dot(x, W_h) + theta_h
A_h = np.maximum(0, Z_h)
Z_o = np.dot(A_h, W_o) + theta_o
y_hat = np.maximum(0, Z_o)
loss = 0.5 * (y_hat - y) ** 2

# Backward pass
d_o = (y_hat - y) * (1.0 if Z_o > 0 else 0.0)
dW_o = np.outer(A_h, d_o).flatten()
d_h = np.dot(d_o, W_o) * (Z_h > 0).astype(float)
dW_h = np.outer(x, d_h)

# Weight updates
W_o_updated = W_o - lr * dW_o
W_h_updated = W_h - lr * dW_h

print("--- Forward Pass Results ---")
print(f"Predicted Output (y_hat) : {y_hat:.4f}")
print(f"Loss                     : {loss:.4f}\n")

print("--- Backward Pass Gradients ---")
print(f"Output Weight Gradients (dW_o)  :\n{np.round(dW_o, 6)}")
print(f"Hidden Weight Gradients (dW_h)  :\n{np.round(dW_h, 6)}\n")

print("--- Updated Parameters ---")
print(f"Updated Output Weights (W_o)    :\n{np.round(W_o_updated, 6)}")
print(f"Updated Hidden Weights (W_h)    :\n{np.round(W_h_updated, 6)}")
```

**Key Takeaways**

Working through manual forward and backward passes makes the internal mechanics of neural networks clear instead of treating them as black boxes. During forward propagation, the ReLU activation zeroed out the first hidden neuron because its pre-activation output was negative, which effectively blocked that unit from contributing to the prediction. In backpropagation, this same zeroing act propagated backward through the derivative of ReLU, which completely blocked gradient flow to the first neuron's weights. As a result, only the active paths received non-zero weight updates, proving how non-differentiable or zero-gradient regions in activation functions directly dictate parameter updates during gradient descent.
