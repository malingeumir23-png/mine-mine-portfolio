# Laboratory Task 2: Forward Pass & Error Computation

## Task Overview
Perform a single forward pass through a multi-layer neural network with ReLU activations and compute the prediction error.

---

## 1. Given Parameters

- **Input vector ($x$)**: $[1, 0, 1]^T$
- **Target output ($y$)**: $[1]$
- **Activation function**: $f(Z) = \max(0, Z)$ (ReLU)

### Weights and Biases
- **Hidden Unit Weights ($W_h$)**:
  $$\begin{bmatrix} 0.2 & -0.3 \\ 0.4 & 0.1 \\ -0.5 & 0.2 \end{bmatrix}$$
- **Output Unit Weights ($W_o$)**:
  $$\begin{bmatrix} -0.3 \\ -0.2 \end{bmatrix}$$
- **Biases ($\theta$)**: $\theta_1 = -0.4$, $\theta_2 = 0.2$, $\theta_3 = 0.1$

---

## 2. Calculation Steps

### Step 1: Hidden Layer Activations
$$Z_{h1} = (1 \cdot 0.2) + (0 \cdot 0.4) + (1 \cdot -0.5) + (-0.4) = -0.7$$
$$Z_{h2} = (1 \cdot -0.3) + (0 \cdot 0.1) + (1 \cdot 0.2) + 0.2 = 0.1$$

Applying ReLU:
$$A_{h1} = \max(0, -0.7) = 0$$
$$A_{h2} = \max(0, 0.1) = 0.1$$

### Step 2: Output Layer Activation
$$Z_o = (0 \cdot -0.3) + (0.1 \cdot -0.2) + 0.1 = 0.08$$
$$\hat{y} = \max(0, 0.08) = 0.08$$

### Step 3: Error Calculation
$$\text{Linear Error }(\delta) = \hat{y} - y = 0.08 - 1 = -0.92$$
$$\text{Squared Error }(E) = \frac{1}{2}(\hat{y} - y)^2 = 0.4232$$


```python
import numpy as np

# Input and target
x = np.array([1, 0, 1])
y = 1

# Weights
W_h = np.array([[0.2, -0.3], [0.4, 0.1], [-0.5, 0.2]])
W_o = np.array([-0.3, -0.2])

# Biases
theta_h1 = -0.4
theta_h2 = 0.2
theta_o = 0.1

# ReLU
def relu(z):
    return np.maximum(0, z)

# Hidden layer
z_h1 = (1 * 0.2) + (0 * 0.4) + (1 * -0.5) + theta_h1
z_h2 = (1 * -0.3) + (0 * 0.1) + (1 * 0.2) + theta_h2
a_h1 = relu(z_h1)
a_h2 = relu(z_h2)

# Output layer
z_o = (a_h1 * -0.3) + (a_h2 * -0.2) + theta_o
y_hat = relu(z_o)

# Error
linear_error = y_hat - y
squared_error = 0.5 * (y_hat - y) ** 2

print(f"Hidden activation A_h1: {a_h1:.4f}")
print(f"Hidden activation A_h2: {a_h2:.4f}")
print(f"Predicted Output (y_hat): {y_hat:.2f}")
print(f"Linear Error: {linear_error:.4f}")
print(f"Squared Error: {squared_error:.4f}")
```

## 3. Final Answer

- **Hidden activation $A_{h1}$** = 0
- **Hidden activation $A_{h2}$** = 0.1
- **Predicted output ($\hat{y}$)** = 0.08
- **Linear error ($\delta$)** = -0.92
- **Squared error ($E$)** = 0.4232

The forward pass produces a prediction of **0.08** while the target is **1**. Therefore, the prediction is below the target, giving a linear error of **-0.92** and a squared error of **0.4232**.


**Key Takeaway**

Neural network produces an output through a forward pass. The input values are first multiplied by their corresponding weights, then the bias is added before applying the ReLU activation. For the first hidden unit, the result was -0.7, so ReLU changed it to 0. For the second hidden unit, the result was 0.1, so it stayed 0.1. These values were then used in the output layer, which gave a predicted output of 0.08. Since the target value was 1, the prediction was still far from the expected result, giving a linear error of -0.92 and a squared error of 0.4232. This helped me understand that the forward pass is basically how the input moves through the network until it produces a prediction, while the error shows how far that prediction is from the target. The error can then be used later when the network learns and adjusts its weights and biases through backpropagation.
